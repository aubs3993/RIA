````python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import config

df = pd.read_parquet(config.CLEAN_PARQUET)
print(df.shape)
df.head()
````

````python
# Coverage of key fields
for col in ['website', 'cco_email', 'aum_total', 'pct_hnw_clients', 'employee_count']:
    if col in df.columns:
        print(f'{col:<25} non-null: {df[col].notna().mean():.1%}')
````

````python
# AUM distribution (log scale)
import numpy as np
aum = df['aum_total'].dropna()
aum = aum[aum > 0]
np.log10(aum).hist(bins=60)
````
